```python
# show all airline with their frequenceis
data.airline.value_counts()
```

```python
## show all airlines with their number of flights in Horizontal Bar charts
data.airline.value_counts(ascending=True).plot.barh( color=['lightgreen','lightblue'] )
plt.title('Airlines with frequencies')
plt.xlabel("Number of flights")
plt.ylabel("Airlines")
plt.show() ## to remove those extra lines
```

### Q2. Show bar graphs representing the departure time & arrival time

```python
data.head(2)
```

```python
## showing detarture time along with the flights
data.departure_time.value_counts()
```

```python
## showing arrival time for flights
data.arrival_time.value_counts()
```

```python
# showing the departure time and arrival time for the flights with their counts

plt.figure(figsize=(16,4))

# for departure time bar graph
plt.subplot(1,2,1)

plt.bar( data.departure_time.value_counts().index , data.departure_time.value_counts().values,color=['r','b'] )

plt.subplot(1,2,2)
plt.bar(data.arrival_time.value_counts().index, data.arrival_time.value_counts().values, color=['g','y'])

plt.show()
```

### Q3. Show bar graph for representing the source cith & destination city

```python
data.head(2)
```

```python
## showing source_city time for flights

data.source_city.value_counts()
```

```python
## showing destination_city time for flights

data.destination_city.value_counts()
```

```python
## show bar graph for destination_city & source_city for flight within thir counts

plt.figure(figsize=(16,4))

plt.subplot(1,2,1)

plt.barh(data.source_city.value_counts().index , data.source_city.value_counts().values, color=['m','g'])
plt.title('Source cities with number of flights')
plt.xlabel('cities')
plt.ylabel('No. of flights')

plt.subplot(1,2,2)

plt.barh(data.destination_city.value_counts().index , data.destination_city.value_counts().values, color=['y', 'r'])
plt.title('Destination cities with number of flights')
plt.xlabel('cities')
plt.ylabel('No. of flights')

plt.show()
```

### Q4. Does price varies with airlines ?

```python
data.head(2)
```

```python
## grouping the airlines and checking their mean price

data.groupby('airline')['price'].mean()
```

```python
# drawing categortical plot for showing the mean tickets price for each ailine
# we will use seaborn 

sns.catplot(x = 'airline', y = 'price', kind = 'bar',  palette = 'rocket',  data = data,  hue = 'class')
plt.show()
```
